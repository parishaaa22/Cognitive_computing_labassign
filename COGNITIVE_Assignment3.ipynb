```python
# Q6

import pandas as pd

# Create employee dataset
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data)

print(employees)
```

```python
# Q6(a)

print("Shape of DataFrame:", employees.shape)
```

```python
employees = employees.sort_values(by="Salary", ascending=False)

print(employees)
```

```python
employees = employees.sort_values(by="Salary", ascending=False)

print(employees)
```

```python
def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(performance)

print(employees)
```

```python
print(employees.isnull().sum())
```

```python
employees = employees.rename(columns={"Employee_ID": "ID"})

print(employees)
```

```python
print(employees[employees["Years_of_Experience"] > 5])
```

```python
print(employees[employees["Department"] == "IT"])
```

```python
employees["Tax"] = employees["Salary"] * 0.10

print(employees)
```

```python
employees.to_csv("employees_modified.csv", index=False)

print("File saved successfully!")
```
