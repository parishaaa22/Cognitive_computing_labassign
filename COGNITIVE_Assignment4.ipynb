```python
import pandas as pd

roll_number = "1024170366"

# Last two digits
last_two_digits = [int(d) for d in roll_number[-2:]]

categories = ["billing", "account", "general"]

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

# Personalized entries
personalized_entries = []

for d in last_two_digits:
    category = categories[d % 3]

    if category == "billing":
        personalized_entries.append({
            "question": "how can i check my fee payment status",
            "answer": "You can check your fee payment status in the student portal.",
            "keywords": "fee payment status",
            "category": category
        })

# Create DataFrame
df = pd.DataFrame(fixed_entries + personalized_entries)

print(df)
```

```python
def score_query(query):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        score = len(query_words & keywords)

        if score > 0:
            results.append((score, row["question"], row["answer"]))

    results.sort(reverse=True)

    for score, question, answer in results:
        print("Confidence:", score)
        print("Question:", question)
        print("Answer:", answer)
        print()

# Example
score_query("fee payment")
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]

print(same_category("billing", df))
```

```python
def score_query_tie(query):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        score = len(query_words & keywords)

        if score > 0:
            results.append((score, row["question"], row["answer"]))

    results.sort(reverse=True)

    if results:
        highest_score = results[0][0]

        print("Highest confidence:", highest_score)

        for score, question, answer in results:
            if score == highest_score:
                print("Question:", question)
                print("Answer:", answer)
                print()
    else:
        print("No matching entries found.")


# Query producing a tie
score_query_tie("fee")

# Query not producing a tie
score_query_tie("password")
```
