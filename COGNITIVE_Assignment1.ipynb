```python
# Q5
my_dict = {
    "name": "Minie",
    "roll_no": "1024170366",
    "branch": "CSE",
    "age": 19,
    "city": "Patiala"
}
# (i) Rename city to location
my_dict["location"] = my_dict.pop("city")
print("After renaming city:", my_dict)
#(ii) Add CGPA
my_dict["cgpa"] = 8.5
print("After adding CGPA:", my_dict)
#(iii) Increase age by 1
my_dict["age"] += 1
print("After increasing age:", my_dict)
#(iv) Delete branch using pop() and del
dict1 = my_dict.copy()
dict2 = my_dict.copy()
removed_value = dict1.pop("branch")
print("Using pop():", dict1)
print("Removed value:", removed_value)
del dict2["branch"]
print("Using del:", dict2)
# pop() returns the removed value, while del simply deletes the key.
# (v) Iterate using items()
for key, value in my_dict.items():
    print(key, "=>", value)
#(vi) Check whether email exists
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email not found")
# (vii) Create friend dictionary
friend_dict = {
    "name": "Friend",
    "roll_no": "1024170367",
    "branch": "CSE",
    "age": 20,
    "city": "Delhi"
}
merged_dict = {**my_dict, **friend_dict}
print("Merged dictionary:", merged_dict)
# Values from friend_dict win when both dictionaries have the same key.
# (viii) Dictionary containing only string values
string_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}
print("String values dictionary:", string_dict)
```
