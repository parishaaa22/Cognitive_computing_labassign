```python
#Q1
import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter the value of M: "))

x = np.linspace(-10, 10, 100)

y1 = M * x**2
y2 = M * np.sin(x)

plt.plot(x, y1, label="M*x²", linestyle="-")
plt.plot(x, y2, label="M*sin(x)", linestyle="--")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Mathematical Functions")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Subject": ["Maths", "Python", "DBMS", "DAA", "Statistics"],
    "Score": [85, 92, 78, 88, 95]
}

df = pd.DataFrame(data)

print(df)

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    x="Subject",
    y="Score",
    data=df,
    palette="viridis"
)

for container in ax.containers:
    ax.bar_label(container)

plt.title("Subject-wise Scores")
plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.grid(axis="y")

plt.show()
```

```python
#Q3
import numpy as np
import matplotlib.pyplot as plt

# Roll number as seed
np.random.seed(1024170366)

# Generate 50 random values
data = np.random.randn(50)

x = np.arange(1, 51)

# Cumulative sum
cumulative_sum = np.cumsum(data)

# Random noise
noise = np.random.randn(50)

# 2 x 2 subplot
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# 1. Cumulative sum line plot
axes[0, 0].plot(x, cumulative_sum)
axes[0, 0].set_title("Cumulative Sum")
axes[0, 0].set_xlabel("Index")
axes[0, 0].set_ylabel("Cumulative Sum")
axes[0, 0].grid(True)

# 2. Scatter plot with random noise
axes[0, 1].scatter(x, data + noise)
axes[0, 1].set_title("Scatter Plot with Random Noise")
axes[0, 1].set_xlabel("Index")
axes[0, 1].set_ylabel("Value")
axes[0, 1].grid(True)

# 3. Original random data
axes[1, 0].plot(x, data)
axes[1, 0].set_title("Random Data")
axes[1, 0].set_xlabel("Index")
axes[1, 0].set_ylabel("Value")
axes[1, 0].grid(True)

# 4. Histogram
axes[1, 1].hist(data)
axes[1, 1].set_title("Distribution of Random Data")
axes[1, 1].set_xlabel("Value")
axes[1, 1].set_ylabel("Frequency")
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
```

```python
#Q4(i)
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/AnjulaMehto/MCA/main/company_sales_data.csv"

sales = pd.read_csv(url)

plt.figure(figsize=(10, 5))

sns.lineplot(
    x="month_number",
    y="total_profit",
    data=sales,
    marker="o"
)

plt.title("Total Profit of All Months")
plt.xlabel("Month Number")
plt.ylabel("Total Profit")
plt.grid(True)

plt.show()
```

```python
#Q4(ii)
products = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer"
]

plt.figure(figsize=(12, 6))

for product in products:
    sns.lineplot(
        x="month_number",
        y=product,
        data=sales,
        marker="o",
        label=product
    )

plt.title("Monthly Product Sales")
plt.xlabel("Month Number")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#Q4(iii)
plt.figure(figsize=(14, 7))

sales_melted = sales.melt(
    id_vars="month_number",
    var_name="Feature",
    value_name="Value"
)

sns.barplot(
    x="Feature",
    y="Value",
    data=sales_melted
)

plt.title("Bar Chart for All Features")
plt.xlabel("Features")
plt.ylabel("Values")
plt.xticks(rotation=45)
plt.grid(axis="y")

plt.show()
```
