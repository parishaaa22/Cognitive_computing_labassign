```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
#Q1(a)
result = arr + 2
print(result)
#Q1(b)
result = arr * 3
print(result)
#Q1(c)
result = arr / 2
print(result)
```

```python
#Q2(a)
import numpy as np
arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed array:", arr[::-1])
#Q2(b)(i)
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
print("Most frequent value:", most_frequent)
print("Indices:", np.where(x == most_frequent)[0])
#Q2(b)(ii)
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
print("Most frequent value:", most_frequent)
print("Indices:", np.where(y == most_frequent)[0])
```

```python
#Q4
import numpy as np
# Create 25 evenly spaced numbers from 10 to 100
ucs420_minie = np.linspace(10, 100, 25)
print("Array:", ucs420_minie)
print("Dimensions:", ucs420_minie.ndim)
print("Shape:", ucs420_minie.shape)
print("Total elements:", ucs420_minie.size)
print("Data type:", ucs420_minie.dtype)
print("Total bytes:", ucs420_minie.nbytes)
# Transpose using reshape()
transpose_array = ucs420_minie.reshape(25, 1)
print("Transpose using reshape():")
print(transpose_array)
# Transpose using T
print("Transpose using T:")
print(ucs420_minie.T)
```

```python
#Q5
import numpy as np
ucs420_parisha = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])
print("Mean:", np.mean(ucs420_parisha))
print("Median:", np.median(ucs420_parisha))
print("Maximum:", np.max(ucs420_parisha))
print("Minimum:", np.min(ucs420_parisha))
print("Unique elements:", np.unique(ucs420_parisha))
# Reshape to 4 rows and 3 columns
reshaped_ucs420_parisha = ucs420_parisha.reshape(4, 3)
print("Reshaped array:")
print(reshaped_ucs420_parisha)
# Resize to 2 rows and 3 columns
resized_ucs420_parisha = np.resize(ucs420_parisha, (2, 3))
print("Resized array:")
print(resized_ucs420_parisha)
```
